## 0. Concepto Clave: ¿Cómo aprende la máquina?

Antes de tocar código, debemos distinguir las dos grandes familias del Machine Learning:

### 🅰️ Aprendizaje Supervisado (Con Profesor)
Tenemos los datos ($X$) y también tenemos **la respuesta correcta** ($y$).
*   **Ejemplo:** Enseñar a un niño con tarjetas (Flashcards). Le enseñas la foto de un gato ($X$) y le dices "Esto es un Gato" ($y$).
*   **Objetivo:** Predecir la respuesta correcta para datos nuevos.
*   **Tipos:**
    *   **Regresión:** La respuesta es un número (Precio de casa).
    *   **Clasificación:** La respuesta es una categoría (Sí/No, Gato/Perro).
*   *Nota: Este Sprint se centra 100% en esto.*

### 🅱️ Aprendizaje No Supervisado (Sin Profesor)
Tenemos los datos ($X$), pero **NO tenemos la respuesta** ($y$). No hay etiquetas.
*   **Ejemplo:** Darle a un niño un cubo lleno de piezas de LEGO mezcladas. Nadie le dice qué hacer, pero él solo empieza a agruparlas por colores o tamaños.
*   **Objetivo:** Descubrir patrones ocultos o estructuras.
*   **Tipos:**
    *   **Clustering:** Agrupar datos similares (Clientes VIP vs Normales).
    *   *Nota: Esto lo veremos en el Sprint 2.4.*

# 🧠 Teoría Bloque A: El Paradigma Scikit-Learn
**Sprint 2.1 - Metodología del Machine Learning**

En IA, casi todos los problemas se resuelven con el mismo esquema mental, independientemente del algoritmo que uses.
La librería estándar para esto es **Scikit-Learn** (`sklearn`).

### 🎯 Objetivos:
1.  **Datos:** Entender la diferencia entre **Features ($X$)** y **Target ($y$)**.
2.  **Dimensiones:** Por qué $X$ siempre es una Tabla (2D) e $y$ es una Columna (1D).
3.  **El Flujo Sagrado:** Importar -> Instanciar -> Entrenar (`.fit`) -> Predecir (`.predict`).

## 1. Definición de Datos: X e y

En el Aprendizaje Supervisado, enseñamos con ejemplos.
*   **$X$ (Features/Características):** Los datos que tiene el modelo para "ver" (Horas de estudio, Asistencia).
*   **$y$ (Target/Objetivo):** Lo que queremos adivinar (¿Aprueba? 0=No, 1=Sí).

Vamos a crear un dataset pequeño de estudiantes.

```python
import pandas as pd

# Datos históricos de estudiantes
df = pd.DataFrame({
    'horas_estudio': [1, 2, 3, 5, 8, 10, 12],  # X
    'asistencia':    [50, 60, 50, 80, 90, 95, 100], # X (otra feature)
    'aprueba':       [0, 0, 0, 0, 1, 1, 1]     # y (Target: 0=No, 1=Sí)
})

print("--- Datos Históricos ---")
display(df)

# PASO CRÍTICO: SEPARAR X e y
# X: Features (Datos de entrada). Ahora usamos DOS columnas.
# y: Target (Lo que queremos predecir).
X = df[['horas_estudio', 'asistencia']] # TODO: Añade 'asistencia'
y = df['aprueba']                    # TODO: columna 'aprueba'

print(f"Forma de X: {X.shape} (Filas, Columnas)") # Ahora dirá (7, 2)
print(f"Forma de y: {y.shape} (Filas,)")
display(X.head(3))
```

## 2. El Flujo: Fit & Predict

Todos los modelos de `sklearn` funcionan igual. Son clases (objetos) vacíos que aprenden cuando les das datos.

Vamos a usar una **Regresión Logística** (aunque suene raro, se usa para clasificar Sí/No).

```python
# 1. IMPORTAR
from sklearn.linear_model import LogisticRegression

# 2. INSTANCIAR (Crear el cerebro vacío)
modelo = LogisticRegression()

# 3. ENTRENAR (FIT)
# Aquí es donde ocurre la magia. El modelo busca la fórmula matemática.
print("🧠 Entrenando modelo...")
modelo.fit(X, y) # TODO: Usa el método .fit()

print("¡Modelo entrenado!")
# Podemos ver qué ha aprendido (La pendiente y el corte)
print(f"Coeficiente (Importancia de estudiar): {modelo.coef_}")
```

## 3. Predicción (El Examen)

Ahora que el modelo ha aprendido, vamos a preguntarle por un caso nuevo.
**IMPORTANTE:** Para evitar *warnings* y errores, los datos nuevos deben tener **el mismo formato** que los datos de entrenamiento (mismos nombres de columna).

```python
# Llega un estudiante nuevo: Pepito.
# Pepito ha estudiado 6 horas... ¡pero también necesitamos saber su asistencia!
# Supongamos que fue al 80% de las clases.

# IMPORTANTE: El DataFrame nuevo debe tener EXACTAMENTE las mismas columnas que la X original.
pepito = pd.DataFrame({
    'horas_estudio': [6], # TODO: Pon 6 horas
    'asistencia':    [80]  # TODO: Pon 80 de asistencia
})

# 4. PREDECIR (PREDICT)
prediccion = modelo.predict(pepito) # TODO: Usa .predict()

print(f"Predicción para Pepito: {prediccion[0]}")
# Fíjate: Si pones pocas horas pero mucha asistencia, ¿qué pasa?
# 0 = Suspende, 1 = Aprueba

# BONUS: PREDECIR PROBABILIDAD
# A veces no queremos un Sí/No rotundo, sino un % de seguridad.
probabilidad = modelo.predict_proba(pepito)
print(f"Probabilidad de suspender vs aprobar: {probabilidad[0]}")
```
